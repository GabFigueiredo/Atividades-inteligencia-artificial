### Exercício 19

Crie dois arrays de números inteiros aleatórios de tamanho 20 e calcule o ângulo entre eles (produto escalar normalizado).

```python
import numpy as np

# Cria o primeiro array com 20 números inteiros aleatórios entre 1 e 100.
array1 = np.random.randint(1, 101, 20)

# Cria o segundo array com 20 números inteiros aleatórios entre 1 e 100.
array2 = np.random.randint(1, 101, 20)

# Calcula o produto escalar entre os dois arrays.
produto_escalar = np.dot(array1, array2)

# Calcula o tamanho (norma) do primeiro array.
norma1 = np.linalg.norm(array1)

# Calcula o tamanho (norma) do segundo array.
norma2 = np.linalg.norm(array2)

# Calcula o cosseno do ângulo entre os dois arrays.
cos_angulo = produto_escalar / (norma1 * norma2)

# Calcula o ângulo em radianos.
angulo = np.arccos(cos_angulo)

# Converte o ângulo de radianos para graus.
angulo_graus = np.degrees(angulo)

# Mostra o primeiro array.
print("Array 1:")
print(array1)

# Mostra o segundo array.
print("\nArray 2:")
print(array2)

# Mostra o ângulo entre os arrays em graus.
print("\nÂngulo entre os arrays:")
print(angulo_graus)
```

### Exercício 1

Calcule a integral da função f(x) = x² sin(x) no intervalo de 0 a π.

```python
import numpy as np
from scipy.integrate import quad

# Define a função f(x) = x² * sen(x)
def f(x):
    return x**2 * np.sin(x)

# Define o limite inferior da integral
limite_inferior = 0

# Define o limite superior da integral
limite_superior = np.pi

# Calcula numericamente a integral usando o método quad
resultado, erro = quad(f, limite_inferior, limite_superior)

# Mostra o resultado da integral
print("Resultado da integral:", resultado)

# Mostra uma estimativa do erro numérico
print("Estimativa do erro:", erro)
```

### Exercício 3

Encontre as raízes da função f(x) = x⁴ − 3x³ + 2 utilizando SciPy.

```python
import numpy as np
from scipy.optimize import root

# Define a função f(x) = x⁴ - 3x³ + 2
def f(x):
    return x**4 - 3*x**3 + 2

# Define diferentes valores iniciais para procurar as raízes
chutes_iniciais = [-1, 0, 1, 2, 3, 4]

# Cria uma lista vazia para armazenar as raízes encontradas
raizes = []

# Percorre cada chute inicial
for chute in chutes_iniciais:

    # Tenta encontrar uma raiz começando pelo chute atual
    resultado = root(f, chute)

    # Verifica se o método encontrou uma solução com sucesso
    if resultado.success:

        # Armazena o valor da raiz encontrada
        raiz = resultado.x[0]

        # Verifica se essa raiz ainda não foi armazenada
        if not any(np.isclose(raiz, raiz_existente) for raiz_existente in raizes):

            # Adiciona a nova raiz à lista
            raizes.append(raiz)

# Ordena as raízes do menor para o maior valor
raizes.sort()

# Mostra as raízes encontradas
print("Raízes da função:")

# Percorre todas as raízes encontradas
for raiz in raizes:

    # Mostra cada raiz com 6 casas decimais
    print(f"{raiz:.6f}")
```

### Exercício 4

Use a função de otimização de SciPy para maximizar f(x) = −x² + 4x + 1.

```python
import numpy as np
from scipy.optimize import minimize

# Define a função que queremos maximizar
def f(x):
    return -x[0]**2 + 4*x[0] + 1

# Define uma nova função que será minimizada
# Usamos o negativo porque o minimize procura o menor valor
def funcao_para_minimizar(x):
    return -f(x)

# Define um valor inicial para começar a busca
x_inicial = [0]

# Realiza a otimização a partir do valor inicial
resultado = minimize(funcao_para_minimizar, x_inicial)

# Obtém o valor de x encontrado pelo método
x_maximo = resultado.x[0]

# Calcula o valor máximo da função nesse ponto
valor_maximo = f([x_maximo])

# Mostra o valor de x onde ocorre o máximo
print("Valor de x:", x_maximo)

# Mostra o valor máximo da função
print("Valor máximo de f(x):", valor_maximo)
```

### Exercício 5

Calcule a transformada de Fourier de um sinal composto de sin(2πt) e cos(4πt).

```python
import numpy as np
from scipy.fft import fft, fftfreq

# Define a quantidade de pontos que serão usados no sinal
N = 1000

# Define o intervalo de tempo do sinal
tempo = np.linspace(0, 1, N)

# Define a frequência de amostragem do sinal
frequencia_amostragem = N

# Cria o sinal composto pela senoide e pela cossenoide
sinal = np.sin(2 * np.pi * tempo) + np.cos(4 * np.pi * tempo)

# Calcula a Transformada de Fourier do sinal
transformada = fft(sinal)

# Calcula as frequências correspondentes aos valores da transformada
frequencias = fftfreq(N, 1 / frequencia_amostragem)

# Calcula a intensidade de cada frequência
intensidade = np.abs(transformada)

# Seleciona somente a primeira metade das frequências
frequencias_positivas = frequencias[:N // 2]

# Seleciona as intensidades correspondentes às frequências positivas
intensidades_positivas = intensidade[:N // 2]

# Encontra a frequência que possui a maior intensidade
indice_maior = np.argmax(intensidades_positivas)

# Obtém o valor da frequência principal
frequencia_principal = frequencias_positivas[indice_maior]

# Mostra a frequência principal encontrada
print("Frequência principal:", frequencia_principal, "Hz")
```

### Exercício 10

Calcule a integral dupla da função f(x, y) = xy + x² no intervalo [0, 2] para x e [0, 1] para y.

```python
import numpy as np
from scipy.integrate import dblquad

# Define a função que será integrada
def f(y, x):
    return x * y + x**2

# Define o limite inferior de y
limite_y_inferior = 0

# Define o limite superior de y
limite_y_superior = 1

# Define o limite inferior de x
limite_x_inferior = 0

# Define o limite superior de x
limite_x_superior = 2

# Calcula numericamente a integral dupla
resultado, erro = dblquad(
    f,
    limite_x_inferior,
    limite_x_superior,
    lambda x: limite_y_inferior,
    lambda x: limite_y_superior
)

# Mostra o resultado da integral
print("Resultado da integral:", resultado)

# Mostra a estimativa do erro numérico
print("Estimativa do erro:", erro)
```

```python
import numpy as np
from scipy.optimize import minimize

# Define a função que queremos minimizar
def f(variaveis):
    # Obtém o valor de x
    x = variaveis[0]

    # Obtém o valor de y
    y = variaveis[1]

    # Calcula o valor da função
    return (x - 2)**2 + (y - 3)**2

# Define os valores iniciais de x e y
chute_inicial = [0, 0]

# Realiza a otimização começando pelos valores iniciais
resultado = minimize(f, chute_inicial)

# Obtém o valor de x encontrado
x_minimo = resultado.x[0]

# Obtém o valor de y encontrado
y_minimo = resultado.x[1]

# Mostra o valor de x no ponto mínimo
print("Valor de x:", x_minimo)

# Mostra o valor de y no ponto mínimo
print("Valor de y:", y_minimo)

# Mostra o valor mínimo da função
print("Valor mínimo de f(x, y):", resultado.fun)
```

### Exercício 16

Resolva a equação diferencial de segunda ordem d²y/dx² + y = 0, com condições iniciais y(0) = 1 e y′(0) = 0.

```python
import numpy as np
from scipy.integrate import solve_ivp

# Define o sistema de duas equações de primeira ordem
def equacao(x, y):

    # Define a primeira equação: y1' = y2
    dy1 = y[1]

    # Define a segunda equação: y2' = -y1
    dy2 = -y[0]

    # Retorna as duas derivadas
    return [dy1, dy2]

# Define o intervalo de x de 0 até 10
intervalo = (0, 10)

# Define as condições iniciais y(0) = 1 e y'(0) = 0
condicoes_iniciais = [1, 0]

# Resolve numericamente o sistema de equações diferenciais
solucao = solve_ivp(equacao, intervalo, condicoes_iniciais)

# Mostra os valores de x calculados
print("Valores de x:")
print(solucao.t)

# Mostra os valores de y calculados
print("Valores de y:")
print(solucao.y[0])
```

### Exercício 19

Realize a transformada de Fourier de um sinal discreto composto de senos e cossenos.

```python
import numpy as np
from scipy.fft import fft, fftfreq

# Define a quantidade de pontos do sinal
N = 100

# Define a frequência de amostragem
frequencia_amostragem = 100

# Cria os pontos de tempo do sinal
tempo = np.arange(N) / frequencia_amostragem

# Cria o sinal composto por um seno e um cosseno
sinal = np.sin(2 * np.pi * tempo) + np.cos(4 * np.pi * tempo)

# Calcula a Transformada de Fourier do sinal
transformada = fft(sinal)

# Calcula as frequências correspondentes aos resultados da transformada
frequencias = fftfreq(N, 1 / frequencia_amostragem)

# Calcula a intensidade de cada frequência
intensidade = np.abs(transformada)

# Seleciona somente as frequências positivas
frequencias_positivas = frequencias[:N // 2]

# Seleciona somente as intensidades das frequências positivas
intensidades_positivas = intensidade[:N // 2]

# Encontra as posições das duas maiores intensidades
indices = np.argsort(intensidades_positivas)[-2:]

# Mostra as frequências encontradas
print("Frequências principais:")

# Percorre as posições das maiores intensidades
for indice in indices:

    # Mostra a frequência correspondente
    print(frequencias_positivas[indice], "Hz")
```

Crie um gráfico de barras empilhadas para três categorias com três grupos de valores.

```python
import matplotlib.pyplot as plt

# Cria as três categorias do gráfico
categorias = ["Categoria A", "Categoria B", "Categoria C"]

# Define os valores dos grupos
grupo1 = [10, 20, 15]
grupo2 = [15, 10, 20]
grupo3 = [5, 15, 10]

# Cria as barras do primeiro grupo
plt.bar(categorias, grupo1, label="Grupo 1")

# Cria as barras do segundo grupo empilhadas sobre o primeiro grupo
plt.bar(categorias, grupo2, bottom=grupo1, label="Grupo 2")

# Calcula a altura acumulada dos dois primeiros grupos
base_grupo3 = [grupo1[i] + grupo2[i] for i in range(len(categorias))]

# Cria as barras do terceiro grupo empilhadas sobre os dois primeiros
plt.bar(categorias, grupo3, bottom=base_grupo3, label="Grupo 3")

# Define o título do gráfico
plt.title("Gráfico de barras empilhadas")

# Define o nome do eixo x
plt.xlabel("Categorias")

# Define o nome do eixo y
plt.ylabel("Valores")

# Exibe a legenda dos grupos
plt.legend()

# Exibe o gráfico
plt.show()
```

### Exercício 18

Crie um gráfico de barras com valores positivos e negativos e use diferentes cores para cada tipo de valor.

```python
import matplotlib.pyplot as plt

# Cria as categorias do gráfico
categorias = ["A", "B", "C", "D", "E", "F"]

# Cria os valores positivos e negativos
valores = [20, -15, 30, -10, 25, -20]

# Cria uma lista de cores verificando se cada valor é positivo ou negativo
cores = ["blue" if valor >= 0 else "red" for valor in valores]

# Cria o gráfico de barras usando as categorias, valores e cores
plt.bar(categorias, valores, color=cores)

# Adiciona uma linha horizontal no valor zero
plt.axhline(0, color="black", linewidth=0.8)

# Define o título do gráfico
plt.title("Valores positivos e negativos")

# Define o nome do eixo x
plt.xlabel("Categorias")

# Define o nome do eixo y
plt.ylabel("Valores")

# Exibe o gráfico
plt.show()
```
